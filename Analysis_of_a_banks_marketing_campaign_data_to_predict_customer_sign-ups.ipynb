```python
import pandas as pd   
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
path = "Bank_Marketing_Campaign.csv"
df = pd.read_csv(path)
df.head
```

```python
print(df.info())
```

```python
print(df.shape)
print('---------------------------')
print(df.isnull().sum())
print('---------------------------')
null_value_percentage = (df.isnull().sum()/df.shape[0])*100
print(null_value_percentage.sort_values(ascending= False))
```

```python
df.nunique()
```

```python
cat_cols = [df.columns[i] 
            for i in range(1,df.shape[1]-1) 
            if df.iloc[:,i].dtype=="str" ]
print(cat_cols)
```

```python
fig ,axes = plt.subplots(3,3,figsize=(25,20))
axes = [ax for axes_rows in axes for ax in axes_rows]
for i, column in enumerate(df [cat_cols]):
    print(f'i: {i} \n column: {column} \n df[column].value_counts()[::-1] {df[column].value_counts()[::-1]} \n ---------------------')
    
    df[column].value_counts()[::-1].plot(kind='pie', ax=axes[i], title=column, autopct='%.0f %%', fontsize=12)
```

```python
fig ,axes = plt.subplots(3,3,figsize=(25,20))
axes = [ax for axes_rows in axes for ax in axes_rows]
for i, column in enumerate(df [cat_cols]):
    df[column].value_counts()[::-1].plot(kind='barh', ax=axes[i], title=column, fontsize=12)
```
